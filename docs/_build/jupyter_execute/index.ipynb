## Introducción

### Contexto

Hay tres fases reconocidas en una partida de ajedrez: **apertura**, donde predomina el desarrollo de las piezas y el control del centro; **medio juego**, donde se producen maniobras de defensa y ataque contra el rey del oponente o sus debilidades; y **final**, donde, generalmente después de varios intercambios de piezas, la promoción de peones se convierte en el tema dominante.

Uno de los aspectos más fascinantes del ajedrez es el estudio de los finales, donde el número de piezas se reduce y el juego se vuelve más teórico y preciso. Uno de ellos es el de **Rey+Torre vs Rey+Peón**.

---

### El dataset

El dataset **KRKPA7** (**K**ing + **R**ook vs **K**ing + **P**awn on **a7**) contiene las posiciones del final de ajedrez de 3,196 partidas, correspondientes al final **Rey + Torre (blancas) vs Rey + Peón (negras)** en la posición a7, es decir, con el peón negro a punto de coronar (el tablero se mira desde la perspectiva del negro). Fueron generadas y descritas originalmente por **Alen D. Shapiro** en su tesis doctoral (1983).

Cada posición está representada por **36 variables categóricas** que describen características específicas del tablero.

La variable objetivo (`class`) indica si las blancas pueden ganar (`won`) o no (`nowin`).

### Observación sobre los códigos

`g` y `w` representan el mismo concepto general (acción o característica de las blancas), pero su uso depende de la variable específica:

- **`w` (White)**: Se usa cuando la variable indica directamente una acción o posición de las blancas.
- **`g` (Good)**: Se usa en la variable `dwipd` para indicar una posición favorable para las blancas (good/ganancia).
